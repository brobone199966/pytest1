实验1：基本运算、数据类型与字符串

```python
# 1. 分别计算并输出结果及类型
print(8 / 3, type(8 / 3))   # 普通除法，结果为浮点数
print(8 // 3, type(8 // 3)) # 整除（向下取整），结果为整数
print(8 % 3, type(8 % 3))   # 取模（求余数），结果为整数
print(2 ** 7, type(2 ** 7)) # 幂运算，结果为整数

# 2. 创建不同类型数据并检查
a = 100
b = 3.14
c = "Python"
d = True
print(type(a), type(b), type(c), type(d))

# 3. 类型转换与异常捕获
print(int("123"), float("3.14"), int(3.9)) # int(3.9) 会直接截断小数部分变为 3
try:
    int("goat")
except ValueError as e:
    print(f"捕获到异常: {e}。原因: 字符串 'goat' 无法被转换为整数。")

# 4. 字符串操作
text = " Python Data Analysis "
text_stripped = text.strip() # 去除首尾空白
print(text_stripped.upper()) # 全大写
print(text_stripped.lower()) # 全小写
print(text_stripped[0:6])    # 切片
words = text_stripped.split() # 拆分
print(words)
joined_text = "-".join(words) # 连接
print(joined_text)

# 5. 拼接、重复与不可变性
s1, s2 = "Hello", "World"
print(s1 + " " + s2) # 拼接
print(s1 * 3)        # 重复

try:
    s1[0] = 'h' # 尝试修改单个字符
except TypeError as e:
    print(f"报错: {e}。字符串是不可变的。")
# 构造新字符串
new_s = 'h' + s1[1:]
print(f"新字符串: {new_s}")
```

实验2：函数、条件、循环与递归

```python
# 1. score_level 函数
def score_level(score):
    if score < 0 or score > 100:
        raise ValueError("分数必须在0到100之间")
    if score >= 90: return "优秀"
    elif score >= 80: return "良好"
    elif score >= 60: return "及格"
    else: return "不及格"
print(score_level(85))

# 2. 循环求和
sum_for = sum(range(1, 101))
sum_while = 0
i = 1
while i <= 100:
    sum_while += i
    i += 1
print(f"for循环: {sum_for}, while循环: {sum_while}")

# 3. 递归版和迭代版阶乘
def fact_rec(n):
    if n < 0: raise ValueError("必须为非负整数")
    return 1 if n == 0 else n * fact_rec(n - 1)

def fact_iter(n):
    if n < 0: raise ValueError("必须为非负整数")
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

for n in [0, 1, 5, 10]:
    print(f"n={n}, 递归: {fact_rec(n)}, 迭代: {fact_iter(n)}")

# 4. summary 函数
def summary(values):
    if not values: raise ValueError("列表不能为空")
    return (min(values), max(values), sum(values) / len(values))

min_val, max_val, avg_val = summary([85, 58, 92, 76, 60, 89])
print(f"最小值:{min_val}, 最大值:{max_val}, 平均值:{avg_val:.2f}")

# 5. sum_all 函数
def sum_all(*args):
    return sum(args)
print(sum_all(), sum_all(5), sum_all(1, 2, 3, 4, 5))
```

实验3：列表、字典、元组与基础统计

```python
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

# 1. zip 构建字典
score_dict = dict(zip(names, scores))
for name, score in score_dict.items():
    print(f"{name}: {score}")

# 2. enumerate 输出序号
for index, (name, score) in enumerate(zip(names, scores), start=1):
    print(f"序号{index}: {name} - {score}分")

# 3. 列表推导式
passed_names = [name for name, score in zip(names, scores) if score >= 60]
squared_scores = [score ** 2 for score in scores]
print("及格名单:", passed_names)
print("成绩平方:", squared_scores)

# 4. 基础统计
count = len(scores)
avg = sum(scores) / count
pass_rate = (len(passed_names) / count) * 100
print(f"人数:{count}, 平均分:{avg:.2f}, 最高分:{max(scores)}, 最低分:{min(scores)}, 及格率:{pass_rate:.2f}%")

# 5. 拷贝对比
demo_scores = [85, 58, 92]
alias = demo_scores
copied = demo_scores.copy()
alias[0] = 99
print(f"原列表被修改了: {demo_scores}") # alias和demo指向同一个
print(f"副本未被修改: {copied}")

# 6. 元组与zip不等长
t = ("Alice", 85)
try:
    t[0] = "Bob"
except TypeError as e:
    print(f"元组元素不能修改: {e}")
print("不等长zip:", list(zip([1, 2, 3], ['a', 'b']))) # 以短的为准
```

实验4：文件读写、异常处理与模块

```python
import os
import pickle
import student_utils # 导入你刚才写的模块

# 1. 目录操作
print("当前工作目录:", os.getcwd())
data_dir = os.path.join(os.getcwd(), "exp01_data")
os.makedirs(data_dir, exist_ok=True)

# 2. 写入数据
file_path = os.path.join(data_dir, "students.txt")
content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete"""
with open(file_path, "w", encoding="utf-8") as f:
    f.write(content)

# 3 & 4. 按行处理与异常捕获
valid_records = []
invalid_count = 0
with open(file_path, "r", encoding="utf-8") as f:
    lines = f.readlines()

for line_num, line in enumerate(lines[1:], start=2):
    try:
        record = student_utils.parse_student(line)
        valid_records.append(record)
    except ValueError as e:
        invalid_count += 1
        print(f"行{line_num}无效: {e}")

print(f"有效记录: {len(valid_records)}条, 无效记录: {invalid_count}条")

# 5. 写入 report.txt (w 和 a 模式)
report_path = os.path.join(data_dir, "report.txt")
scores_list = [r["score"] for r in valid_records]
avg_score = sum(scores_list) / len(scores_list)
pass_rt = len([s for s in scores_list if s >= 60]) / len(scores_list) * 100

with open(report_path, "w", encoding="utf-8") as f:
    f.write(f"有效记录数: {len(valid_records)}\n")
    f.write(f"无效记录数: {invalid_count}\n")
    f.write(f"平均分: {avg_score:.2f}\n")
    f.write(f"及格率: {pass_rt:.2f}%\n")

with open(report_path, "a", encoding="utf-8") as f:
    f.write("--- 追加测试 ---\n") # 'w'模式会清空原文件，'a'模式会在末尾追加

# 6. pickle 存取
pkl_path = os.path.join(data_dir, "students.pkl")
with open(pkl_path, "wb") as f:
    pickle.dump(valid_records, f)
with open(pkl_path, "rb") as f:
    loaded_records = pickle.load(f)
print(f"pickle读取一致: {loaded_records == valid_records}")

# 7. 测试模块调用
print("有效输入测试:", student_utils.parse_student("Tom, 99"))
try:
    student_utils.parse_student("Jack, 120")
except ValueError as e:
    print("无效输入测试:", e)

# 8. 捕获文件不存在异常
try:
    with open("no_such_file.txt", "r") as f:
        pass
except FileNotFoundError as e:
    print(f"成功捕获异常: {e}")
```

实验5：类、继承与对象复制

```python
import copy

# 1. Student 类
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score
    
    def is_passed(self):
        return self.score >= 60
    
    def __str__(self):
        return f"Student: {self.name}, 分数: {self.score}, 及格: {self.is_passed()}"

# 2. 从实验4数据创建对象
students_objs = [Student(r["name"], r["score"]) for r in valid_records]
for s in students_objs:
    print(s)

# 3. GraduateStudent 类继承
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic
    
    def __str__(self):
        return f"GraduateStudent: {self.name}, 分数: {self.score}, 课题: {self.research_topic}, 及格: {self.is_passed()}"

grad = GraduateStudent("Alice", 85, "AI")
print(grad)

# 4. 深拷贝与浅拷贝
def create_obj():
    return {"name": "Alice", "scores": [85, 90]}

# 直接赋值 (完全指向同一块内存)
o1 = create_obj()
a1 = o1
a1["scores"].append(100)
print(f"直接赋值 - 原对象: {o1['scores']}")

# 浅拷贝 (copy.copy)
o2 = create_obj()
a2 = copy.copy(o2)
a2["scores"].append(100)
print(f"浅拷贝 - 原对象: {o2['scores']} (内部嵌套列表被连带修改)")

# 深拷贝 (copy.deepcopy)
o3 = create_obj()
a3 = copy.deepcopy(o3)
a3["scores"].append(100)
print(f"深拷贝 - 原对象: {o3['scores']} (不受影响)")
```
